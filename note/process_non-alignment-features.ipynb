```julia
DIR_DATA = "/net/dali/home/barton/kais/Neutralization-Completion/data_Kai/CATNAP_reodo"
fname_features_V1 = "$(DIR_DATA)/var_reg_char_25-12-30_14-8_V1.csv"
fname_features_V2 = "$(DIR_DATA)/var_reg_char_25-12-30_14-8_V2.csv"
fname_features_V1V2 = "$(DIR_DATA)/var_reg_char_25-12-30_14-8_V1+V2.csv"
fname_features_V3 = "$(DIR_DATA)/var_reg_char_25-12-30_14-8_V3.csv"
fname_features_V4 = "$(DIR_DATA)/var_reg_char_25-12-30_14-8_V4.csv"
fname_features_V5 = "$(DIR_DATA)/var_reg_char_25-12-30_14-8_V5.csv"

csv_features_raw_V1 = CSV.read(fname_features_V1, DataFrame);
csv_features_raw_V2 = CSV.read(fname_features_V2, DataFrame);
csv_features_raw_V1V2 = CSV.read(fname_features_V1V2, DataFrame);
csv_features_raw_V3 = CSV.read(fname_features_V3, DataFrame);
csv_features_raw_V4 = CSV.read(fname_features_V4, DataFrame);
csv_features_raw_V5 = CSV.read(fname_features_V5, DataFrame);
```

```julia
# Merge all V region features into a single DataFrame
# Start with V1 and keep Name column
csv_features_merged = select(csv_features_raw_V1, :Name, 
    :Length => :Length_V1, 
    Symbol("Glyco(Nx[ST])") => Symbol("Glyco(Nx[ST])_V1"),
    Symbol("NP[ST]") => Symbol("NP[ST]_V1"),
    Symbol("NN[ST][ST]") => Symbol("NN[ST][ST]_V1"),
    :Charge => :Charge_V1)

# Join V2
csv_features_merged = leftjoin(csv_features_merged, 
    select(csv_features_raw_V2, :Name, 
        :Length => :Length_V2, 
        Symbol("Glyco(Nx[ST])") => Symbol("Glyco(Nx[ST])_V2"),
        Symbol("NP[ST]") => Symbol("NP[ST]_V2"),
        Symbol("NN[ST][ST]") => Symbol("NN[ST][ST]_V2"),
        :Charge => :Charge_V2), 
    on = :Name)

# Join V1V2
csv_features_merged = leftjoin(csv_features_merged, 
    select(csv_features_raw_V1V2, :Name, 
        :Length => :Length_V1V2, 
        Symbol("Glyco(Nx[ST])") => Symbol("Glyco(Nx[ST])_V1V2"),
        Symbol("NP[ST]") => Symbol("NP[ST]_V1V2"),
        Symbol("NN[ST][ST]") => Symbol("NN[ST][ST]_V1V2"),
        :Charge => :Charge_V1V2), 
    on = :Name)

# Join V3
csv_features_merged = leftjoin(csv_features_merged, 
    select(csv_features_raw_V3, :Name, 
        :Length => :Length_V3, 
        Symbol("Glyco(Nx[ST])") => Symbol("Glyco(Nx[ST])_V3"),
        Symbol("NP[ST]") => Symbol("NP[ST]_V3"),
        Symbol("NN[ST][ST]") => Symbol("NN[ST][ST]_V3"),
        :Charge => :Charge_V3), 
    on = :Name)

# Join V4
csv_features_merged = leftjoin(csv_features_merged, 
    select(csv_features_raw_V4, :Name, 
        :Length => :Length_V4, 
        Symbol("Glyco(Nx[ST])") => Symbol("Glyco(Nx[ST])_V4"),
        Symbol("NP[ST]") => Symbol("NP[ST]_V4"),
        Symbol("NN[ST][ST]") => Symbol("NN[ST][ST]_V4"),
        :Charge => :Charge_V4), 
    on = :Name)

# Join V5
csv_features_merged = leftjoin(csv_features_merged, 
    select(csv_features_raw_V5, :Name, 
        :Length => :Length_V5, 
        Symbol("Glyco(Nx[ST])") => Symbol("Glyco(Nx[ST])_V5"),
        Symbol("NP[ST]") => Symbol("NP[ST]_V5"),
        Symbol("NN[ST][ST]") => Symbol("NN[ST][ST]_V5"),
        :Charge => :Charge_V5), 
    on = :Name);
```

```julia
DIR_PROCESSED = "../processed-data"
CSV.write( "$(DIR_PROCESSED)/var_reg_char_25-12-30_14-8_V1toV5_merged.csv", csv_features_merged);
```
